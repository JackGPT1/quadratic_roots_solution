```python
import math

def quadratic_roots(a,b,c):
  if a!=0:
    Delta = b**2-4*a*c

    if Delta > 0:
      sq = math.sqrt(Delta)
      x_1 = (-b+sq)/(2*a)
      x_2 = (-b-sq)/(2*a)
      print(f"Two different real roots are ")
      print(f"x_1 = {x_1}")
      print(f"x_2 = {x_2}")
    elif Delta == 0:
      sq = math.sqrt(Delta)
      x_1 = (-b+sq)/(2*a)
      x_2 = x_1
      print(f"Two identical real roots are")
      print(f"x_1 = {x_1}")
      print(f"x_2 = {x_2}")
    else:
      x_1 = -b/(2*a) + (math.sqrt(-1*Delta)/(2*a))*1j
      x_2 = -b/(2*a) - (math.sqrt(-1*Delta)/(2*a))*1j
      print(f"Two different complex roots are")
      print(f"x_1 = {x_1}")
      print(f"x_2 = {x_2}")
    return x_1,x_2

  else:
    print("Please input valid non-zero 'a' again!")
    return None
```

```python
a = 20
b = 20
c = 20

root1, root2 = quadratic_roots(a,b,c)
print(f"{root1:.3f}")
print(f"{root2:.3f}")
```
