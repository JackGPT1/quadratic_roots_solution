```python
import math
import matplotlib.pyplot as plt
import numpy as np

def quadratic_roots(a,b,c):
  if a!=0:
    Delta = b**2-4*a*c

    if Delta > 0:
      sq = math.sqrt(Delta)
      x_1 = (-b+sq)/(2*a)
      x_2 = (-b-sq)/(2*a)
      print(f"Two different real roots are ")
      print(f"x_1 = {x_1}")
      print(f"x_2 = {x_2}")

    elif Delta == 0:
      sq = math.sqrt(Delta)
      x_1 = (-b+sq)/(2*a)
      x_2 = x_1
      print(f"Two identical real roots are")
      print(f"x_1 = {x_1}")
      print(f"x_2 = {x_2}")

    else:
      x_1 = -b/(2*a) + (math.sqrt(-1*Delta)/(2*a))*1j
      x_2 = -b/(2*a) - (math.sqrt(-1*Delta)/(2*a))*1j
      print(f"Two different complex roots are")
      print(f"x_1 = {x_1}")
      print(f"x_2 = {x_2}")

    return x_1,x_2

  else:
    print("Please input valid non-zero 'a' again!")
    return None


def quadratic_plot(a,b,c):
    lower_bound = -b/(2*a) - 30
    upper_bound = -b/(2*a) + 30

    x_values = np.linspace(lower_bound,upper_bound)
    y_values = a*x_values**2+b*x_values+c

    plt.plot(x_values,y_values)
    plt.show()
```

```python
a = 2
b = 3
c = 40


root1, root2 = quadratic_roots(a,b,c)
print(f"{root1:.3f}")
print(f"{root2:.3f}")
```

```python
quadratic_plot(a,b,c)
```
